```python
import gradio as gr
from google import genai
from pypdf import PdfReader

client = genai.Client(api_key=API_KEY)

prompts = {
    "Summary": "Summarize these study notes in simple English using short bullet points. "
               "End with 3 key points to remember.",
    "Quiz (10 MCQs)": "Create 10 multiple choice questions from these notes. "
                      "Give 4 options (A-D) for each question. "
                      "Put all the correct answers at the end under 'Answers'.",
    "Flashcards": "Create 10 flashcards from these notes in the format:\n"
                  "Q: question\nA: short answer",
}

def generate(notes, pdf_path, mode):
    text = notes or ""

    # If a PDF is uploaded, read its text
    if pdf_path:
        reader = PdfReader(pdf_path)
        for page in reader.pages:
            text += "\n" + (page.extract_text() or "")

    if not text.strip():
        return "Please paste notes or upload a PDF first."

    try:
        response = client.models.generate_content(
            model="gemini-3.8-flash",  # change if you get 'model not found'
            contents=f"{prompts[mode]}\n\nNOTES:\n{text[:15000]}",
        )
        return response.text
    except Exception as e:
        return f"Something went wrong: {e}"

demo = gr.Interface(
    fn=generate,
    inputs=[
        gr.Textbox(label="Paste your notes here", lines=8),
        gr.File(label="...or upload a PDF", type="filepath", file_types=[".pdf"]),
        gr.Radio(["Summary", "Quiz (10 MCQs)", "Flashcards"], value="Summary", label="What do you want?"),
    ],
    outputs=gr.Markdown(label="Result"),
    title="📚Study Helper",
    description="Paste notes or upload a PDF. Get a summary, quiz, or flashcards.",
)

demo.launch(share=True)
```
